## 1. Import Library

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## 2. Request Halaman & Cek Status Code

```python
url = "https://www.detik.com/sumbagsel/berita/d-8371084/50-kata-kata-bijak-hari-ini-terbaru-seru-dan-penuh-arti"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}

page = requests.get(url, headers=headers)
page.status_code
```

## 3. Lihat Potongan HTML Mentah

```python
print(page.text[:500])
```

## 4. Parsing HTML dengan BeautifulSoup

```python
soup = BeautifulSoup(page.text, "html.parser")
soup.title
```

## 5. Eksplorasi Tag Dasar

```python
soup.title.text
```

```python
soup.find("h1")
```

## 6. `find` vs `find_all`

```python
# find -> cuma 1 hasil (paragraf pertama yang ditemukan di seluruh halaman)
soup.find("p")
```

```python
# find_all -> semua hasil, kita intip 3 pertama saja
soup.find_all("p")[:3]
```

## 7. Ambil Judul Artikel

```python
judul = soup.find("h1").get_text(strip=True)
judul
```

## 8. Ambil Metadata Tambahan (Kategori & Tanggal)

```python
kategori_tag = soup.find("div", class_="breadcrumb")
kategori = kategori_tag.get_text(strip=True) if kategori_tag else "Tidak ditemukan"
kategori
```

```python
tanggal_tag = soup.find("div", class_="detail__date")
tanggal = tanggal_tag.get_text(strip=True) if tanggal_tag else "Tidak ditemukan"
tanggal
```

## 9. Ambil Isi Artikel (Paragraf & List)

```python
konten_div = soup.find("div", class_="detail__body-text")

if konten_div:
    paragraf = konten_div.find_all(["p", "li"])
    hasil_teks = [p.get_text(strip=True) for p in paragraf if p.get_text(strip=True)]
else:
    hasil_teks = []
    print("Konten tidak ditemukan, cek ulang struktur HTML-nya (class bisa berubah sewaktu-waktu)")

len(hasil_teks)
```

## 10. Tampilkan Tiap Paragraf dengan Nomor Urut

```python
for i, teks in enumerate(hasil_teks, start=1):
    print(f"{i}. {teks}")
```

## 11. Ubah Hasil Scraping Menjadi DataFrame

```python
df = pd.DataFrame({
    "No": range(1, len(hasil_teks) + 1),
    "Isi Paragraf": hasil_teks
})
df
```
