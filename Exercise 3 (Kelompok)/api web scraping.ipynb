## 1. Import Library

```python
import requests
import pandas as pd
import json
```

## 2. Request Endpoint /posts & Cek Status Code

```python
api_url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(api_url)
response.status_code
```

## 3. Lihat Struktur Data JSON

```python
data = response.json()  # otomatis parsing JSON jadi list of dict
data[0]
```

## 4. Tampilkan 5 Data Pertama

```python
for item in data[:5]:
    print(f"ID: {item['id']} | Judul: {item['title']}")
```

## 5. Ubah Seluruh Data Menjadi DataFrame

```python
df_posts = pd.DataFrame(data)
df_posts.head(10)
```

## 6. Ambil Data dari Endpoint Terkait

```python
post_id = 1
url_comments = f"https://jsonplaceholder.typicode.com/posts/{post_id}/comments"
comments = requests.get(url_comments).json()
comments[:2]
```

## 7. Ubah Data Comments Menjadi DataFrame

```python
df_comments = pd.DataFrame(comments)
df_comments
```

## 8. Gunakan Query Parameter untuk Filter Data

```python
response_filtered = requests.get(
    "https://jsonplaceholder.typicode.com/comments",
    params={"postId": 1}
)
response_filtered.json()[:2]
```

## 9. Loop untuk Ambil Ringkasan Beberapa Post Sekaligus

```python
ringkasan = []

for pid in range(1, 6):
    post = requests.get(f"https://jsonplaceholder.typicode.com/posts/{pid}").json()
    komentar = requests.get(f"https://jsonplaceholder.typicode.com/posts/{pid}/comments").json()
    ringkasan.append({
        "postId": pid,
        "title": post["title"],
        "jumlah_komentar": len(komentar)
    })

ringkasan
```

## 10. Ubah Ringkasan Menjadi DataFrame

```python
df_ringkasan = pd.DataFrame(ringkasan)
df_ringkasan
```

## 11. Simpan Hasil Scraping ke File JSON

```python
with open("hasil scraping posts.json", "w") as f:
    json.dump(data, f, indent=4)

print("Tersimpan sebagai hasil scraping posts.json")
```
